12. Аналіз виживання залежно від статі: Обчисліть відсоток виживання для кожної статі. Чи була різниця у виживанні між чоловіками та жінками?

Головна причина в тому, що дітей евакуювали першочергово, що суттєво підвищило відсоток виживання серед наймолодших пасажирів.

15.	Розподіліть пасажирів на групи за рівнями тарифів (Fare) і обчисліть рівень виживання для кожної групи. Як тариф впливав на шанси виживання (дати відповідь)?

```python
# Розподіляємо вартість квитка (Fare) на 4 квантильні групи (Низька, Середня, Вище середнього, Висока)
df['FareGroup'] = pd.qcut(df['Fare'], q=4, labels=['Низький', 'Середній', 'Вище середнього', 'Високий'])

# Обчислюємо відсоток виживання для кожної групи
survival_by_fare = df.groupby('FareGroup', observed=False)['Survived'].mean() * 100
print(survival_by_fare)
```

Так, тариф суттєво впливав на шанси виживання: чим вищою була ціна квитка (тариф Fare), тим вищим був відсоток виживання. Пасажири з категорії з найвищими тарифами мали значно більше шансів врятуватися (понад 50-60%), оскільки зазвичай це були пасажири 1-го класу, каюти яких розташовувалися вище та ближче до шлюпочної палуби.

```python
mean_fare_by_pclass = df.groupby('Pclass')['Fare'].mean()
print(mean_fare_by_pclass)
```
