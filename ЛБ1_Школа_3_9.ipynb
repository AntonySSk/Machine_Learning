```text
     [3, 4]]
```

### Завдання 23
Для вектора:


обчисліть його евклідову норму.

````python
# Код студента
import numpy as np

# Креирање на векторот
v = np.array([3, 4])

# Пресметување на Евклидовата норма (должина на векторот)
norm = np.linalg.norm(v)

# Нормализација на векторот (вектор со единична должина)
normalized_v = v / norm

print("Евклидова норма:", norm)
print("Нормализиран вектор:", normalized_v)
````

````python
# Код студента
import numpy as np

# Створення векторів
a = np.array([1, 2])
b = np.array([2, 1])

# Обчислення скалярного добутку та норм векторів
dot_prod = np.dot(a, b)
norm_a = np.linalg.norm(a)
norm_b = np.linalg.norm(b)

# Обчислення косинуса кута між векторами
cos_angle = dot_prod / (norm_a * norm_b)

# Знаходження кута у радіанах та переведення в градуси
angle_rad = np.arccos(cos_angle)
angle_deg = np.degrees(angle_rad)

print("Косинус кута:", cos_angle)
print("Кут у градусах:", angle_deg)
````

Нормалізуйте **кожен стовпець** матриці:

```text
A = [[10, 20, 30],
     [20, 30, 40],
     [30, 40, 50]]
```

за формулою:

\[
z = \frac{x - \mu}{\sigma}
\]

де `μ` — середнє значення стовпця, `σ` — стандартне відхилення стовпця.

Не використовуйте цикл `for`.

````python
# Код студента
import numpy as np

# Створення матриці A
A = np.array([
    [10, 20, 30],
    [20, 30, 40],
    [30, 40, 50]
])

# Середнє значення та стандартне відхилення для кожного стовпця (axis=0)
mu = np.mean(A, axis=0)
sigma = np.std(A, axis=0)

# Нормалізація за допомогою broadcasting (без циклу for)
Z = (A - mu) / sigma

print("Середнє значення стовпців (mu):", mu)
print("Стандартне відхилення стовпців (sigma):", sigma)
print("\nНормалізована матриця (Z):\n", Z)
````
